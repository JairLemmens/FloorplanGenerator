```python
import floorplan_backend

outer = [[0.0, 0.0], [120.0, 0.0], [120.0, 40.0],
         [50.0, 40.0], [50.0, 100.0], [0.0, 100.0]]
field_samples = [[25.0, 20.0, 1.0, 0.0], [95.0, 75.0, 0.0, 1.0]]
regions = floorplan_backend.segment_domain(outer, field_samples=field_samples)

print(f"Regions: {len(regions)}")
for i, rings in enumerate(regions):
    print(f"Region {i}: {rings}")
```

```python

import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

fig, ax = plt.subplots()
for i, _polygon in enumerate(regions):
    vertices = _polygon[0]

    ax.add_patch(Polygon(
        vertices,
        closed=True,
        facecolor=f'C{i}',
        edgecolor='black',
        alpha=0.4
    ))
    
ax.set_aspect('equal')
ax.autoscale_view()
ax.grid(True)
```
